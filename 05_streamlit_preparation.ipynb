```python
import pandas as pd
import numpy as np
import pickle
import joblib

from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
priority_model = load_model(
    "models/priority_lstm.keras"
)

with open(
    "models/priority_tokenizer.pkl",
    "rb"
) as f:
    priority_tokenizer = pickle.load(f)

with open(
    "models/priority_encoder.pkl",
    "rb"
) as f:
    priority_encoder = pickle.load(f)

print("Priority Model Loaded")
```

```python
category_model = load_model(
    "models/priority_category_lstm.keras"
)

with open(
    "models/category_tokenizer.pkl",
    "rb"
) as f:
    category_tokenizer = pickle.load(f)

with open(
    "models/category_encoder.pkl",
    "rb"
) as f:
    category_encoder = pickle.load(f)

print("Category Model Loaded")
```

```python
rf_model = joblib.load(
    "models/resolution_time_rf.pkl"
)

print("Regression Model Loaded")
```

```python
sample_text = """
payment deducted twice refund not received
"""

seq = priority_tokenizer.texts_to_sequences(
    [sample_text]
)

pad = pad_sequences(
    seq,
    maxlen=100,
    padding="post"
)

pred = priority_model.predict(pad)

pred_class = np.argmax(pred)

priority = priority_encoder.inverse_transform(
    [pred_class]
)

print("Predicted Priority:")
print(priority[0])
```

```python
seq = category_tokenizer.texts_to_sequences(
    [sample_text]
)

pad = pad_sequences(
    seq,
    maxlen=100,
    padding="post"
)

pred = category_model.predict(pad)

pred_class = np.argmax(pred)

category = category_encoder.inverse_transform(
    [pred_class]
)

print("Predicted Category:")
print(category[0])
```

```python
sample_df = pd.DataFrame(
    [[
        30,
        24,
        5,
        4,
        10,
        6,
        120,
        150,
        3
    ]],
    columns=rf_model.feature_names_in_
)

prediction = rf_model.predict(
    sample_df
)

print(
    "Predicted Resolution Hours:",
    round(prediction[0], 2)
)
```
